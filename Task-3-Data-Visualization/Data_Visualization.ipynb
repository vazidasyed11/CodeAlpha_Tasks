```python
import pandas as pd

data = {
    "Student": [
        "Anu", "Ravi", "Priya", "Rahul", "Sneha",
        "Arjun", "Divya", "Kiran", "Pooja", "Naveen",
        "Swathi", "Akhil", "Meena", "Suresh", "Lakshmi",
        "Varun", "Ayesha", "Rohit", "Kavya", "Manoj"
    ],

    "Study Hours": [
        5, 3, 7, 4, 8,
        6, 2, 5, 7, 4,
        9, 3, 6, 5, 8,
        4, 7, 2, 6, 5
    ],

    "Attendance": [
        85, 70, 92, 76, 95,
        88, 65, 82, 90, 73,
        97, 68, 86, 80, 93,
        75, 89, 62, 84, 78
    ],

    "Assignment Score": [
        78, 65, 88, 72, 92,
        80, 60, 75, 85, 68,
        95, 64, 82, 76, 90,
        70, 86, 58, 81, 74
    ],

    "Exam Score": [
        82, 68, 91, 74, 94,
        85, 62, 79, 89, 71,
        96, 66, 84, 78, 92,
        73, 88, 60, 83, 76
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Study Hours",
    y="Exam Score"
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.show()
```

Study Hours vs Exam Score

Observation:
Students who spend more hours studying generally have higher exam scores.

Insight:
Regular study time appears to have a positive relationship with exam performance.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Exam Score"
)

plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")
plt.show()
```

Attendance vs Exam Score

Observation:
Students with higher attendance generally have better exam scores.

Insight:
Good attendance may help students understand lessons better and improve their examination performance.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Assignment Score",
    y="Exam Score"
)

plt.title("Assignment Score vs Exam Score")
plt.xlabel("Assignment Score")
plt.ylabel("Exam Score")
plt.show()
```

Assignment Score vs Exam Score

Observation:
Students with higher assignment scores generally also have higher exam scores.

Insight:
Good performance in assignments can indicate better preparation for examinations.

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=df,
    x="Student",
    y="Exam Score"
)

plt.title("Exam Scores of 20 Students")
plt.xlabel("Students")
plt.ylabel("Exam Score")
plt.xticks(rotation=45)

plt.show()
```

Exam Scores of 20 Students

Observation:
The bar chart shows differences in the performance of the 20 students.

Insight:
Students such as Swathi, Sneha, and Lakshmi have very high exam scores, while Rohit and Divya have comparatively lower scores.

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Graph 1: Study Hours vs Exam Score
sns.scatterplot(
    data=df,
    x="Study Hours",
    y="Exam Score",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Study Hours vs Exam Score")


# Graph 2: Attendance vs Exam Score
sns.scatterplot(
    data=df,
    x="Attendance",
    y="Exam Score",
    ax=axes[0, 1]
)
axes[0, 1].set_title("Attendance vs Exam Score")


# Graph 3: Assignment vs Exam Score
sns.scatterplot(
    data=df,
    x="Assignment Score",
    y="Exam Score",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Assignment Score vs Exam Score")


# Graph 4: Exam Scores
sns.barplot(
    data=df,
    x="Student",
    y="Exam Score",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Exam Scores of Students")
axes[1, 1].tick_params(axis="x", rotation=45)

# Dashboard title
fig.suptitle(
    "Student Performance Data Visualization Dashboard",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

### Student Performance Data Story

The dashboard analyzes the performance of 20 students using study hours, attendance, assignment scores, and exam scores.

The visualizations show that students who spend more time studying generally achieve higher exam scores. Students with better attendance also tend to perform better in examinations. Assignment scores show a similar relationship with exam scores.

The dashboard helps identify high-performing and low-performing students. Teachers can use this information to identify students who may need additional support and encourage better study habits and attendance.

Overall, data visualization makes it easier to understand student performance and helps in making better academic decisions.


### Conclusion

Data visualization converts raw student data into meaningful graphs and charts. Using Python, Pandas, Matplotlib, and Seaborn, the student performance data was analyzed through different visualizations. The analysis helped identify relationships between study hours, attendance, assignment scores, and exam scores. These visualizations make the data easier to understand and support better academic decision-making.
